## 수치 해의 오차

1. Errors in Numerical Solutions and Representation of Numbers on a Computer

1.1. 수치해석 개요 및 오차 분석

- 해석적인 해(analytical solutio, exact)과 수치적인 해(numerical soluttion, approximate)
- [예] 평형 방정식(equations of equilibrium)을 적용하여 구하는 경우, 해석적 해는 못 구하더라도 수치적 해는 근사할 수 있다.
    
    $\mu (mg - F \sin \theta) = F \cos \theta \\
    \rightarrow F = \frac{\mu mg}{\cos \theta + F \ \sin \theta}$

1.2. 컴퓨터의 숫자 표현 및 수치 오차

- 수 표현 : 2진수로 컴퓨터 메모리 저장
- 반올림 오차(Round-off Error) : 유한한 비트 수로 실수 정확하게 저장 불가
- 절단 오차(Truncation Error) : 수학적 근사 절차 유한한 반복 계산 또는 근사식 대체 시 발생
- 오차 평가 : 참 오차(True Error)와 상대적 참 오차(True Relative Error) 비교 평가
