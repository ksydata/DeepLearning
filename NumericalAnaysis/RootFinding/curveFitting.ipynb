[개념]
- 벡터의 길이(norm) 기반으로 벡터를 비교 / paraboloid
- 데이터 $y_i$와 함수값 $f(x_i)$의 차이는 +- 0가 된다.
- 1-Norm : $$E = \sum_{i=1}^{n} |r_i| = \sum_{i=1}^{n} |y_i - (a_1 x_i + a_0)| = \|r\|_1$$
- 2-Norm : $$E(a_0, a_1) = \sum_{i=1}^{n} r_i^2 = \sum_{i=1}^{n} \left[ y_i - (a_1 x_i + a_0) \right]^2 = \|r\|_2^2$$

[사례-1]
- $\begin{align}
\frac{\partial E}{\partial a_1}
  &= -2 \sum_{i=1}^{n} (y_i - a_1 x_i - a_0)\, x_i
   = -2 \left[ \mathbf{x}^T \mathbf{y} - (\mathbf{x}^T \mathbf{x})\, a_1 - (\mathbf{x}^T \mathbf{1})\, a_0 \right] = 0, \\
\frac{\partial E}{\partial a_0}
  &= -2 \sum_{i=1}^{n} (y_i - a_1 x_i - a_0)
   = -2 \left[ \mathbf{1}^T \mathbf{y} - (\mathbf{1}^T \mathbf{x})\, a_1 - (\mathbf{1}^T \mathbf{1})\, a_0 \right] = 0.
\end{align}$

[사례-2] 어떻게 비선형 방정식을 선형 변환할까, 로그를 이용한 접근
- 거듭제곱 함수 : $y = b x^m \rightarrow \ln y = m \ln x + \ln b$
- 지수 함수 : $y = b e^{mx} \rightarrow \ln y = mx + \ln b$
- 역수 함수 : $y = \frac{1}{mx+b} \rightarrow \frac{1}{y} = mx + b$
