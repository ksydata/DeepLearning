## 8. Numerical Differentiation (미분, rate of change of a quantity)

- 우리는 언제 수치 미분이 필요할까. 
- 접근법
  - Finite difference approximation
  - Differentiating analytically after curve fitting

#### 8.1. 유한 차분법(finite difference approximating derivatives)

- 연속적인 함수의 도함수(미분값)를 이산적인(discretized) 점들의 함수값을 이용해 근사하는 수치해석 기법
  - forward difference : $f'(a) = lim_{h \rightarrow 0} \frac{f(a+h) - f(a)}{h} \rightarrow \frac{df}{dx} \approx \frac{f(x_{i+1} - f(x_{i}))}{x_{i+1} - x_{i}}$ 
- 
  - backward difference : $f'(a) = lim_{h \rightarrow 0} \frac{f(a) - f(a-h)}{h} \rightarrow \frac{df}{dx} \approx \frac{f(x_{i} - f(x_{i-1}))}{x_{i} - x_{i-1}}$ 

  - **central difference : $f'(a) = lim_{h \rightarrow 0} \frac{f(a+h) - f(a-h)}{2h} \rightarrow \frac{df}{dx} \approx \frac{f(x_{i+1} - f(x_{i-1}))}{x_{i+1} - x_{i-1}}$**

#### 8.2. 테일러 급수 전개 (Taylor Series Expansion)

- 테일러 급수란 어떤 점 $x_{i}$ 근처에서 함수 $f(x)$를 다항식으로 근사하는 방법
- 즉, $x_{i}$에서의 접선으로 $f(x)$를 근사하며, $f(x) = 0$이 되는 $x$를 찾는 것
- $f(x) ≈ f(x_i) + f'(x_i) * (x - x_i) \\
  \ 0 = f(x_i) + f'(x_i) * (x - x_i) \\ 
  \ → x - x_i = -f(x_i) / f'(x_i) \\ 
  \ → x = x_i - f(x_i) / f'(x_i)$
- 접선이라는 1차식으로 잠깐 근사해서 그 1차식의 근을 구하고, 그걸 다음 추측값으로 쓰는 것을 반복하는 방법

- $f(x) = f(x_{0}) + \frac{f'(x_{0})(x-x_{1})}{1!} +  \frac{f''(x_{0})(x-x_{2})^2}{2!} + \cdots$
  - 좌변을 i번 미분하여 $x_{0}$ 대입한 값과 $x_{0}$의 g를 i차 미분한 도함수와 같음
  - $g''(x) = f''(x_{0}) + f'''(x_{0}) + \frac{1}{2!}f''''(x_{0}) + \cdots$ 
  - 미분을 많이 할수록 값이 작아지며(분자), 분모는 커짐. 따라서 차수가 높아질수록 영향력이 점점 줄어듬
- $\Sigma_{i=0}^{n} \frac{}{} + \Sigma_{i=n+1}^{\infty} \frac{}{}$
  - [예] $f(x) = sin(x)$
    - $f'(x) = cos(x), f''(x) = -sin(x), f'''(x) = -cos(x), f^{(4)}(x) = sin(x), f^{(5)}(x) = cos(x)$
    - $f(x_{0} + h) = \Sigma_{i=0}^{n} \frac{f^{n}(x_{0}+h)}{i!} h^{i} = O(h^{n+1})$
    - $R_k(h) = \frac{f(k+1)(ξ)}{(k+1)!} h^{k+1}$ If h is halved, the error shirinks to (1/2)k+1 of the previous one.
