## 신경망 초기 모델, 퍼셉트론

```python
import numpy as np
```

```python
def perceptron(x_1, x_2):
    """퍼셉트론 구현(인공 신경망)"""
    w_1, w_2, b = 1.0, 1.0, -1.5
    sum = x_1 * w_1 + x_2 * w_2 + b
    # 계단 함수 형태의 퍼셉트론에서의 활성화 함수
    # sum = np.dot(W: np.array, X: np.array) + b

    if sum > 0. : return 1 
    else : return 0
```

```python
print( perceptron(0, 0) ,
       perceptron(1, 0) ,
       perceptron(0, 1) , 
       perceptron(1, 1) )
```

퍼셉트론 출력 계산

- $y = \begin{cases}
1 & \text{if } (w_1 x_1 + w_2 x_2 + b \geq 0) \\
0 & \text{otherwise} \end{cases}$

- $x1 = 0,  x2 = 0  →  w1x1+w2x2 = 1x0 + 1x0 = 0  →  b = -1.5  →  y = 0 \\
x1 = 1,  x2 = 0  →  w1x1+w2x2 = 1x1 + 1x0 = 1  →  b = -1.5  →  y = 0 \\
x1 = 0,  x2 = 1  →  w1x1+w2x2 = 1x0 + 1x1 = 1  →  b = -1.5  →  y = 0 \\
x1 = 1,  x2 = 1  →  w1x1+w2x2 = 1x1 + 1x1 = 2  →  b = -1.5  →  y = 1$
